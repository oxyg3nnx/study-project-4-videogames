<b>На доработку🤔:</b>

По коду ниже верно, а по аргументации выше - нет. Важно, чтобы код и комментарии сочетались друг с другом.

Просто заполнить региональные продажи нулем будет плохим решением, так слишком сильно исказим данные. Отсутствие информации не говорит об отсутствии продаж. Примерно как отсутствие ценника на товаре в магазине не говорит, что товар бесплатный.


В задании нам предлагается заполнить здесь пропуски средним, но с группировкой по году и платформе. 


Ок, кажется, что все-таки ситуация, что на месте пропусков у нас нулевые продажи, маловероятна. Более вероятна ситуация, что наши данные неполные. Тогда стоит посмотреть, у нас пропуски в региональных продажах идут в годах, когда в основном продавались консоли, а игры к ним были приложениями и стоили примерно одинаково. Игр было мало и многие, кто покупал консоли, покупали большинство игр. Продажи игр внутри одного года и консили были примерно одинаковыми. А вот продажи в среднем по региону без учета платформы и года могли отличаться на порядки.



````python
print("Пропусков ДО заполнения:")
print(df[['eu_sales', 'jp_sales', 'rating']].isna().sum())
print()

# 1. Заполнение пропусков в продажах средним по группе
for col in ['eu_sales', 'jp_sales']:
    # Для каждого уникального сочетания года и платформы
    for (year, platform), group in df.groupby(['year_of_release', 'platform']):
        # Считаем среднее по этой группе
        group_mean = group[col].mean()
        if pd.notna(group_mean):
            # Заполняем пропуски в этой группе
            mask = (df['year_of_release'] == year) & (df['platform'] == platform) & (df[col].isna())
            df.loc[mask, col] = group_mean
    
    # Оставшиеся пропуски заполняем общим средним
    df[col] = df[col].fillna(df[col].mean())

# 2. Заполнение рейтинга
df['rating'] = df['rating'].fillna('Unknown')

print("Пропусков ПОСЛЕ заполнения:")
print(df[['eu_sales', 'jp_sales', 'rating']].isna().sum())
````
